# Common

```python
import shutil
from pathlib import Path

# root containing sample folders
SAMPLE_ROOT_LE = Path(r".../RAW_EXP_DATA/LE Rat (200um thick)/30um MEA")

SAMPLE_ROOT_RCS = Path(r".../RAW_EXP_DATA/RCS Rat (126um thick)/30um MEA")

# root containing saline folders
SALINE_ROOT = Path(r".../RAW_EXP_DATA/Saline")

# target destination
TARGET_ROOT = Path(".../target_folders")

# mapping from data folders to x folders
data_map = {
    "data001": "x1",
    "data002": "x10",
    "data003": "x30",
    "data004": "x50",
    "data005": "x100",
}

# mapping from line folders to line index
line_map = {
    "line_1": "0",
    "line_2": "1",
    "line_3": "2",
}

def copy_source_files_into_destination_structure(data_map, line_map, saline_mapping, data_folders, sample_root=None, saline_root=None, target_root=None):
	# initialize folders if not provided
	if saline_root is None:
	    saline_root = SALINE_ROOT
	if sample_root is None:
		sample_root = SAMPLE_ROOT_LE
	if target_root is None:
		 target_root = TARGET_ROOT

	# if the target folder doesn't exist, create it
	if not target_root.exists():
		target_root.mkdir(parents=True)

	# copy saline data
	for src_name, sample_name, timing in saline_mapping:
		for data_folder, x_folder in data_map.items():
			for line_folder, line_id in line_map.items():

				src = saline_root / src_name / data_folder / line_folder
				dst = (
						target_root
						/ f"{sample_name}-{line_id}"
						/ "raw_data"
						/ f"{x_folder}_{timing}"
				)

				if src.exists():
					if dst.exists():
						shutil.rmtree(dst)
					shutil.copytree(src, dst)

	# copy sample data
	for sample_name in data_folders:
		for line_folder, line_id in line_map.items():

			src = sample_root / sample_name / "data001" / line_folder
			dst = (
					target_root
					/ f"{sample_name}-{line_id}"
					/ "raw_data"
					/ "sample"
			)

			if src.exists():
				if dst.exists():
					shutil.rmtree(dst)
				shutil.copytree(src, dst)
```
